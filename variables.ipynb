# Nested If Statement

```python
num=int(input("Enter any number :"))
if(num<0):
    print("The number is Negative :",num)
elif(num>0):
    if(num<=10):
        print('The number is butween 1 to 10 :',num)
    elif(num<=20):
        print("The number is between 11 to 20 :",num)   
    else:
        print('The number is Greater then 2o :',num)
else:
    print('The number is Zero :',num)        
```

# Exercise If Else Using Time Module

```python
import time
import datetime

current_time = datetime.datetime.now()
print("Current time:", current_time,'\n')

time_stamp=time.strftime('%H :%M :%S')
print("Print Current time :",time_stamp)
time_stamp=int(time.strftime('%H'))
print("Hour  :",time_stamp)
Hour_time_stamp=int(time.strftime('%M '))
print("Minuts  :",time_stamp)
time_stamp=int(time.strftime('%S'))
print("Seconds :",time_stamp)
if(Hour_time_stamp>=7 and Hour_time_stamp<=9):
    print("Good Morning")
elif(Hour_time_stamp>=11 and Hour_time_stamp<=3):
    print("Good Afternoon")
else:
    print("Good Evening")


```

# MatchCase Statement

```python
Day=str(input("Could you tell me what day it is today?"))
match Day:
    case 'monday':
        print('Today my  Calss is Machine Learning.')
    case 'tuesday':
        print('Today my  Calss is Big Data.')
    case 'wednesday':
        print('Today my  Calss is Data Science.')
    case 'thirsday':
        print('Today my  Calss is Deep Learning.')
    case 'friday':
        print('Today my  Calss is Lab.')
    case 'saturday':
        print('Today I will sleep until 12 AM.')
    case 'sunday':
        print('Today I will go for Foodball.')
    case _:
        print('Invalid day entered.')
        
```
